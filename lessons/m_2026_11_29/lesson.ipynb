## 1. Функция без вызова

Функция является объектом: её можно сохранить под другим именем, положить в коллекцию или передать аргументом. Это нужно, когда общий алгоритм сохраняется, а конкретное действие меняется.

Скобки вызывают функцию: `double(3)` — результат вычисления, `double` — сама функция. Если общая функция ожидает действие, ей передаётся функция без скобок. Такой подход уже встречался в sorted(..., key=len).


```python
def double(value):
    return value * 2

def transform(values, operation):
    return [operation(value) for value in values]

alias = double
print(alias(4), transform([1, 2], double))
assert transform([], double) == []
```

```python
l1 = [double, transform]
```

```python
# l1[1]([1, 2], double)
double_new = l1[1]
double_new([1, 2], double)
```

**Практика:** Объяснить, почему transform([1, 2], double(3)) не передаёт нужное действие.


```python
def calculator(fun, num1, num2):
    result = fun(num1, num2)
    print("Результат операции,", result)

def sum_num(num1, num2):
    print("Считается сумма")
    return num1 + num2

def diff(num1, num2):
    print("Считается разность")
    return num1 - num2
```

```python
calculator(sum_num, 2, 5)
```

```python
calculator(diff, 2, 5)
```

## 2. lambda и key

lambda создаёт небольшую безымянную функцию из одного выражения. Её результатом становится значение выражения. Запись удобна для короткого правила, например выбрать второе поле записи для сортировки.

Если правило требует нескольких шагов или отдельного поясняющего имени, def обычно понятнее. Сортировка стабильна: при равных ключах сохраняется порядок входа. Можно вернуть кортеж нескольких полей, чтобы задать дополнительные критерии.


```python
def fun():
    return 5
```

```python
(lambda: 5)()
```

```python
rows = [("чай", 50), ("кофе", 90), ("сок", 50)]
print(sorted(rows, key=lambda row: row[1]))
print(sorted(rows, key=lambda row: (row[1], row[0])))
assert sorted(["aa", "b", "cc"], key=len) == ["b", "aa", "cc"]
```

```python
users = [
    {"name": "Анна", "age": 28},
    {"name": "Иван", "age": 22},
    {"name": "Мария", "age": 34},
]

by_age = sorted(users, key=lambda u: u["name"])
by_age
```

**Практика:** Сравнить сохранение исходного порядка при равной цене и явную сортировку по названию.


## 3. map, filter и reduce

map применяет функцию к каждому элементу, filter оставляет элементы, для которых предикат истинен. Их результаты выдаются последовательно и могут быть исчерпаны, как zip; list делает готовую коллекцию.

Часто обычное включение читается проще, особенно когда нужны одновременно отбор и преобразование. reduce из functools сворачивает последовательность в один результат, передавая накопитель и очередное значение функции. Для обычной суммы sum понятнее reduce; знать назначение полезно для чтения чужого кода.


```python
from functools import reduce
values = [1, -2, 3]
print(list(map(lambda value: value * 2, values)))
print(list(filter(lambda value: value > 0, values)))
print(reduce(lambda total, value: total + value, values, 0))
```

**Практика:** Переписать map и filter списочными включениями.


## 4. any и all

any проверяет, есть ли хотя бы один истинный элемент, all — все ли элементы истинны. Они удобны для составных условий и проверки набора правил.

Для пустого входа any возвращает False, all — True: нет ни одного нарушившего условие элемента. Если по контракту набор обязан быть непустым, это отдельное требование. Функции прекращают обход, как только результат определён.


```python
values = [2, 4, 6]
print(all([value > 0 for value in values]))
print(any([value < 0 for value in values]))
print(all([]), any([]))
```

**Практика:** Сформулировать условие «непустой набор, все значения положительны».
