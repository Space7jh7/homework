# Аннотации типов

**Результат:** читать и писать понятные сигнатуры, не путая описание типов с проверкой входа.


## 1. Описание интерфейса

Аннотация типа описывает ожидаемый тип параметра, переменной или результата. Она помогает читать интерфейс, получать подсказки редактора и находить несоответствия статическим анализом.

У параметра используется двоеточие, у результата — стрелка после скобок. Аннотация не заменяет понятное имя и описание условий вроде положительности числа. Тип int сам по себе не говорит, разрешён ли отрицательный возраст.


```python
def repeat(text: str, count: int) -> str:
    return text * count

message: str = repeat("API", 2)
print(message)
assert message == "APIAPI"
```

**Практика:** Прочитать сигнатуру словами до разбора тела.


## 2. Коллекции и возможное отсутствие

Параметризованные типы описывают элементы: list[int], dict[str, int], tuple[int, int]. Это не создание коллекции, а описание её ожидаемого содержимого.

`int | None` означает результат одного из двух типов. Такой контракт нужен для поиска без совпадения. Тип с None не делает параметр автоматически необязательным при вызове: для пропуска аргумента нужно отдельное значение по умолчанию.


```python
def first_even(values: list[int]) -> int | None:
    for value in values:
        if value % 2 == 0:
            return value
    return None

print(first_even([1, 3]), first_even([1, 4]))
```

**Практика:** Различить «может содержать None» и «аргумент можно не передавать».


## 3. Аннотация не является валидацией

Обычный Python не запрещает вызвать функцию с аргументом другого типа только из-за аннотации. Если тело сможет обработать значение, вызов выполнится; если нет — ошибка возникнет из-за операции в теле.

Проверка внешних данных выполняется отдельным кодом или валидатором, например будущей Pydantic. Статический анализ и runtime-проверка решают разные задачи. Избыточное использование Any, обозначающего неизвестный для анализатора тип, уменьшает пользу аннотаций.


```python
def identity(value: int) -> int:
    return value

result = identity("не число")
print(result)
assert type(result) == str
```

**Практика:** Объяснить, почему функция выполнилась, несмотря на int в сигнатуре.


## 4. Аннотации и документация

Аннотации полезно писать на границах функций и в моделях данных, а не добавлять механически к каждой очевидной локальной переменной. В сложном контракте отдельная строка документации поясняет смысл, ограничения и побочные эффекты.

Словарь `function.__annotations__` позволяет посмотреть объявленные типы. Это диагностическая информация о функции, а не доказательство её корректного поведения. Работу по-прежнему проверяют вызовами и assert.


```python
def total(values: list[int]) -> int:
    """Сумма значений без изменения входа."""
    return sum(values)

print(total.__annotations__)
assert total([1, 2]) == 3
assert total([]) == 0
```

**Практика:** Добавить понятное описание того, изменяется ли вход функции.
