## Задача 2. Среднее

Написать average(values: list[float]) -> float | None. Для пустого списка вернуть None, иначе среднее.


```python
# РЕШЕНИЕ
def average(values: list[float]) -> float | None:
    if len(values) == 0:               #проверяет не пустой ли список
        return None
    #values[0].
    sum_values = sum(values)           #сумма
    len_values = len(values)           #количество
    c = sum_values / len_values       # результат
    return c
```

```python
# Проверка
assert average([]) is None
assert abs(average([1.0, 2.0]) - 1.5) < 1e-9
assert average([0.0]) == 0.0
print("OK — задача 2")
```

## Задача 3. Счётчик слов — повторение

Реализовать frequencies(words: list[str]) -> dict[str, int]. Не менять вход.


```python
# РЕШЕНИЕ
def frequencies(words: list[str]) -> dict[str, int]:
    a = set()                          #множество для словаря
    d = dict()                           
    for word in words:         
        if word not in a:      #проверяет есть ли значение в множестве
            a.add(word)
            d[word] = 1
        else:                          #если есть то к количеству +1
            d[word] += 1
    return d
```

```python
# Проверка
data = ["api", "sql", "api"]
assert frequencies(data) == {"api": 2, "sql": 1}
assert frequencies([]) == {}
assert data == ["api", "sql", "api"]
print("OK — задача 3")
```

## Задача 4. Границы набора

Реализовать bounds(values: list[int]) -> tuple[int | None, int | None]: минимум и максимум, для пустого — (None, None).


```python
# РЕШЕНИЕ
def bounds(values: list[int]) -> tuple[int | None, int | None]:
    a = []                     #пустой список для определения min и max
    if len(values) == 0:       #проверяет не пустой ли список
        return None,None
    for value in values:
        if len(values) == 1:          #если в списке 1 значение
            return value,value
        a.append(value)                #если в списке 2 и более значение
    return min(a), max(a)
```

```python
# Проверка
assert bounds([3, -1, 5]) == (-1, 5)
assert bounds([]) == (None, None)
assert bounds([4]) == (4, 4)
print("OK — задача 4")
```
