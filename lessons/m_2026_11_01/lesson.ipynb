```python
# Операции над множеством

l1 = [1, 2 ,3]
l2 = [3, 4 ,5]
s1 = set(l1)
s2 = set(l2)

print(l1 + l2)
print(s1 | s2)
```

```python
l1 = [1, 2 ,3, 4]
l2 = [3, 4 ,5]
s1 = set(l1)
s2 = set(l2)

print(s1 - s2)
```

```python
l1 = [1, 2 ,3, 4]
l2 = [3, 4, 5]
s1 = set(l1)
s2 = set(l2)

print(s1 & s2)
```

# Связь множества с хешированием

### Каждый элемент в множестве лежит НЕ по индексу, как в списках, а по адресу хеша объекта и если передаются в множество одинаковые объекты, то они будут лежать на том же месте. По факту если в множество приходит объект, который там уже есть, то множество его не будет в себя жобавлять.

```python

s1 = {1,2,3}
def add_to_set(s1, obj):
    if obj in s1:
        print("можем добавить") # будет проверка по hash
    else:
        print("добавлять НЕ нужно")

add_to_set(s1, 1)
add_to_set(s1, 4)
```
