# 0. Повторение

## 0.1. `for` по списку и `range`

`for` берёт элементы по одному и для каждого выполняет тело цикла.
`range` выдаёт последовательность чисел: `range(стоп)`, `range(старт, стоп)`, `range(старт, стоп, шаг)`.
Правая граница **не входит**, как в срезах.

````python
l1 = [1, 2, 3]

for i in l1:
    print(i)
````

````python
for i in range(5):        # 0, 1, 2, 3, 4 - пятёрка не входит
    print(i)

print(list(range(10, 5, -2)))   # [10, 8, 6] - шаг может быть отрицательным
````

`range` сам по себе не список: он выдаёт числа по одному. Чтобы увидеть их все сразу, оборачиваем в `list(...)`.

## 0.2. `while`

`while` повторяет тело, **пока условие истинно**. Нужен, когда заранее неизвестно, сколько будет шагов.
Главное правило: внутри цикла что-то должно менять условие, иначе цикл не закончится никогда.

````python
balance = 1000
months = 0

while balance < 2000:
    balance = balance * 1.1     # +10% в месяц
    months += 1                 # без этой строки мы не узнаем ответ

print(months, round(balance, 2))
````

## 0.3. Кортеж - `tuple`

Кортеж похож на список, но записывается в круглых скобках и **не меняется** после создания.

````python
l1 = [1, 2, 3]
t1 = tuple(l1)        # из списка сделали кортеж
print(type(t1))
print(t1, t1[0], len(t1))     # индексы, len, срезы - как у списка
````

````python
l1[0] = 2             # список менять можно
print(l1)

# t1[0] = 2           # раскомментируй - будет TypeError: кортеж менять нельзя
````

Зачем нужен кортеж, если есть список:

- он **защищает данные**: случайно поменять его нельзя;
- он **занимает меньше памяти**;
- именно кортежем Python возвращает несколько значений из функции (раздел 3.4).

````python
import sys

l1 = list(range(100_000))
t1 = tuple(l1)

print("список: ", sys.getsizeof(l1), "байт")
print("кортеж: ", sys.getsizeof(t1), "байт")

# список держит запас места под будущие append, кортежу запас не нужен
l1.append(1)
print("список после append:", sys.getsizeof(l1), "байт")
````

### Распаковка

Кортеж (и список) можно «разложить» по переменным одной строкой. Переменных должно быть ровно столько, сколько элементов.

````python
person = ("Коля", 15)
name, age = person
print(name)
print(age)

# Распаковка прямо в цикле: список кортежей
results = [("Коля", 12.4), ("Петя", 11.9)]
for name, seconds in results:
    print(f"{name}: {seconds} с")
````

# 1. `break` - выйти из цикла досрочно

## 1.1. Что делает

`break` **немедленно прекращает весь цикл**, оставшиеся элементы не просматриваются.
Выполнение продолжается со строки после цикла.

Типичная задача - поиск: как только нашли нужное, дальше смотреть незачем.

````python
temperatures = [5, 3, 1, -2, 4, -5]

first_frost = None
for t in temperatures:
    print("смотрю", t)
    if t < 0:
        first_frost = t
        break               # нашли первый мороз - дальше не идём

print("первый мороз:", first_frost)
````

Обрати внимание: `-5` даже не напечаталось. Цикл закончился на `-2`.

Почему перед циклом `first_frost = None`: если мороза в списке нет, `break` не сработает,
и переменная должна остаться в осмысленном состоянии «не нашли».

## 1.2. `while True` + `break`

Иногда условие выхода удобнее проверять **в середине** цикла, а не в заголовке.
Тогда пишут бесконечный цикл `while True` и выходят из него через `break`.

````python
# Кидаем «кубик» по заранее известной последовательности, пока не выпадет шестёрка
rolls = [2, 5, 1, 6, 3, 6]

i = 0
while True:
    roll = rolls[i]
    i += 1
    if roll == 6:
        break
    print("выпало", roll, "- кидаем ещё")

print("шестёрка с попытки номер", i)
````

## 1.3. `break` выходит только из своего цикла

Если циклы вложены, `break` прерывает только **ближайший** (внутренний). Внешний продолжает работать.

````python
classes = [
    [5, 4, 2, 5],     # 7А
    [4, 4, 5],        # 7Б
    [3, 2, 2, 5],     # 7В
]

for marks in classes:
    for mark in marks:
        if mark == 2:
            print("в классе есть двойка:", marks)
            break       # прерываем перебор оценок этого класса, но не перебор классов
````

## 1.4. `for ... else`

У цикла может быть ветка `else`. Она выполняется, если цикл **дошёл до конца без `break`**.
Читать её удобно как «если не нашли».

````python
marks = [5, 4, 4, 5]

for mark in marks:
    if mark == 2:
        print("есть двойка")
        break
else:
    print("двоек нет")      # сюда попадаем, только если break не случился
````

# 2. `continue` - пропустить шаг

## 2.1. Что делает

`continue` **заканчивает текущий шаг** и сразу переходит к следующему элементу.
Цикл при этом не прерывается.

Сравни:

| | Что происходит |
|---|---|
| `break` | выйти из цикла совсем |
| `continue` | бросить этот элемент, взять следующий |

````python
numbers = [4, -1, 7, 0, -3, 5]

total = 0
for n in numbers:
    if n <= 0:
        continue            # неположительные пропускаем
    total += n
    print("добавил", n)

print("сумма положительных:", total)
````

## 2.2. Зачем, если есть `if`

То же самое можно написать через `if n > 0:` и весь код внутри него. Пока строк одна-две, разницы нет.
Но когда проверок несколько, `if` вкладываются друг в друга «лесенкой», а `continue` позволяет
отсеять лишнее **в самом начале** и дальше писать без вложенности.

````python
words = ["кот", "", "Пёс", "ёж", "  ", "сова"]

# Лесенка из if
for w in words:
    if w.strip() != "":
        if len(w) > 2:
            print(w.lower())

print("---")

# То же через continue: сначала отсеиваем, потом работаем
for w in words:
    if w.strip() == "":
        continue
    if len(w) <= 2:
        continue
    print(w.lower())
````

## 2.3. Ловушка: `continue` в `while`

В `while` счётчик меняем мы сами. Если `continue` стоит **до** строки `i += 1`, счётчик перестаёт расти,
и цикл становится бесконечным. Этот код запускать **не надо**:

```python
i = 0
while i < 5:
    if i == 2:
        continue    # i навсегда останется 2
    print(i)
    i += 1
```

Правильно - сначала сдвинуть счётчик, потом решать, пропускать ли шаг:

````python
i = 0
while i < 5:
    i += 1
    if i == 3:
        continue
    print(i)
````

# 3. Функции

## 3.1. Зачем

Функция - это кусок кода с именем. Написали один раз, вызываем сколько угодно.

Без функции одинаковый расчёт приходится копировать. Найдёшь ошибку - исправлять придётся во всех копиях.
С функцией - в одном месте.

Мы уже пользовались чужими функциями: `print`, `len`, `sum`, `sorted`. Теперь пишем свои.

````python
def greet(name):            # name - параметр: «пустое место» под значение
    print(f"Привет, {name}!")


greet("Коля")               # "Коля" - аргумент: то, что подставляется в параметр
greet("Аня")
````

Разбор:

- `def` - «определить функцию», дальше имя, в скобках параметры, в конце двоеточие;
- тело функции - с отступом;
- `def` только **создаёт** функцию, код внутри выполняется в момент **вызова** `greet(...)`.

## 3.2. `return` против `print`

Это главное в теме.

- `print` **показывает** значение на экране. Дальше использовать его нельзя.
- `return` **отдаёт** значение тому, кто вызвал функцию. Его можно сохранить в переменную, сложить, передать дальше.

````python
def average_print(marks):
    print(sum(marks) / len(marks))


def average_return(marks):
    return sum(marks) / len(marks)


a = average_print([5, 4, 3])     # на экран вывелось 4.0 ...
print("в переменной:", a)        # ... но в переменной None

b = average_return([5, 4, 3])    # на экран ничего
print("в переменной:", b)        # а значение есть, с ним можно работать
print("округлили:", round(b + 0.333, 1))
````

**Функция без `return` возвращает `None`.** Если результат функции почему-то оказался `None` -
первым делом проверь, не забыт ли `return` и не написан ли `print` вместо него.

Правило: функция, которая **считает**, должна возвращать результат через `return`.
Печатать его - дело того кода, который её вызвал.

## 3.3. `return` сразу завершает функцию

Как только выполнился `return`, функция заканчивается, даже если он внутри цикла.
Поэтому в функциях `return` часто заменяет `break`: нашли - сразу вернули ответ.

````python
def first_negative(numbers):
    for n in numbers:
        if n < 0:
            return n        # нашли - выходим из функции целиком
    return None             # дошли до конца - значит, не нашли


print(first_negative([5, 3, -2, 4, -5]))   # -2
print(first_negative([1, 2, 3]))           # None
````

Сравни с разделом 1.1: не нужна переменная `first_frost`, не нужен `break`, а случай «не нашли» записан
отдельной понятной строкой в конце.

## 3.4. Вернуть несколько значений

Если после `return` перечислить значения через запятую, функция вернёт **кортеж**.
Получить значения можно по индексу или сразу распаковать.

````python
def fun():
    name = "Vova"
    age = 25
    return name, age        # на самом деле возвращается кортеж ("Vova", 25)


result = fun()
print(type(result))
print(result)
print(result[0], result[1])
````

````python
name, age = fun()           # распаковка - так обычно и пишут
print(name)
print(age)
````

````python
def min_max(numbers):
    smallest = numbers[0]
    largest = numbers[0]
    for n in numbers:
        if n < smallest:
            smallest = n
        if n > largest:
            largest = n
    return smallest, largest


low, high = min_max([4, 9, 1, 7])
print(low, high)
````

## 3.5. Значения по умолчанию и именованные аргументы

Параметру можно задать значение **по умолчанию**. Тогда при вызове его можно не указывать.
Параметры со значением по умолчанию пишутся **после** обычных.

````python
def power(base, exponent=2):
    return base ** exponent


print(power(5))             # 25 - exponent взят по умолчанию
print(power(2, 10))         # 1024 - передали явно
print(power(2, exponent=3)) # 8 - именованный аргумент: видно, что за число 3
print(power(exponent=3, base=2))   # с именами порядок неважен
````

Именованные аргументы делают вызов понятным. Сравни `sorted(marks, reverse=True)` и гипотетическое
`sorted(marks, True)` - во втором случае непонятно, что значит `True`.

## 3.6. Локальные переменные

Переменные, созданные **внутри** функции, существуют только пока функция работает.
Снаружи их не видно. Это хорошо: функции не мешают друг другу одинаковыми именами.

````python
def count_fives(marks):
    count = 0               # локальная переменная
    for m in marks:
        if m == 5:
            count += 1
    return count


print(count_fives([5, 4, 5]))
# print(count)              # раскомментируй - будет NameError: снаружи count не существует
````

````python
discount = 10               # переменная снаружи функции


def price_bad(price):
    return price - discount             # читает внешнюю переменную - работает, но плохо


def price_good(price, discount):
    return price - discount             # всё нужное пришло через параметры


print(price_bad(100))
print(price_good(100, 10))
````

`price_bad` молча зависит от переменной снаружи. Поменяешь `discount` где-то в другой ячейке -
изменится результат, и найти причину будет сложно. **Всё, что нужно функции, передавай ей параметрами.**

## 3.7. Функция и список: изменения видны снаружи

Помнишь ловушку `b = a` из прошлого занятия? Передача списка в функцию работает так же:
параметр - это **второе имя того же списка**. Если функция меняет список, он меняется и снаружи.

````python
def add_mark(marks, mark):
    marks.append(mark)      # меняем тот самый список, который передали


my_marks = [5, 4]
add_mark(my_marks, 3)
print(my_marks)             # [5, 4, 3] - список снаружи изменился
````

Иногда это и нужно. Но чаще функция должна **не трогать** входные данные, а вернуть новый результат.
Тогда новый список собирают внутри функции:

````python
def without_twos(marks):
    result = []                 # новый список, исходный не трогаем
    for m in marks:
        if m == 2:
            continue
        result.append(m)
    return result


my_marks = [5, 2, 4, 2, 3]
clean = without_twos(my_marks)
print(clean)                # [5, 4, 3]
print(my_marks)             # [5, 2, 4, 2, 3] - исходный цел
````

## 3.8. Функции вызывают функции

Большую задачу удобно собирать из маленьких функций, каждая из которых делает одно дело.

````python
def average(marks):
    return sum(marks) / len(marks)


def quarter_mark(marks):
    avg = average(marks)
    if avg >= 4.5:
        return 5
    if avg >= 3.5:
        return 4
    if avg >= 2.5:
        return 3
    return 2


print(quarter_mark([5, 5, 4]))      # 5
print(quarter_mark([4, 3, 4, 3]))   # 4 - среднее 3.5
````

Заметь, в `quarter_mark` нет ни одного `elif` и `else`: каждый `return` сразу завершает функцию,
поэтому до следующего `if` дело доходит, только если предыдущие условия не сработали.

# Частые ошибки

- `print` вместо `return` - результат функции оказывается `None`.
- Код после `return` в той же ветке никогда не выполнится.
- `return` внутри цикла с отступом не на том уровне: функция выходит на первом же элементе.
- Забыли вызвать функцию: `average` вместо `average(marks)` - это сама функция, а не результат.
- `continue` в `while` до изменения счётчика - бесконечный цикл.
- Ждать, что `break` выйдет из обоих вложенных циклов. Он выходит только из ближайшего.
- Функция меняет переданный список, хотя от неё ждали новый.
- Функция читает переменные снаружи вместо параметров.

# Памятка

| Что | Как | Что происходит |
|---|---|---|
| Выйти из цикла | `break` | цикл прекращается, дальше код после него |
| Пропустить шаг | `continue` | переход к следующему элементу |
| Ветка «не нашли» | `for ... else` | `else` выполняется, если не было `break` |
| Бесконечный цикл | `while True:` + `break` | выход по условию в середине |
| Создать функцию | `def name(a, b=1):` | `b` - параметр со значением по умолчанию |
| Вернуть результат | `return x` | функция сразу завершается |
| Вернуть несколько | `return a, b` | возвращается кортеж `(a, b)` |
| Получить несколько | `x, y = f()` | распаковка кортежа |
| Именованный аргумент | `f(1, b=5)` | понятно, что передаём |
| Нет `return` | - | функция возвращает `None` |
| Кортеж | `(1, 2)`, `tuple(lst)` | как список, но не меняется и легче |
