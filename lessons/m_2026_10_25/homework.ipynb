## Задача 5. Ценник

Напиши функцию `format_price(price, currency="руб.")`, которая **возвращает** строку:

- `"бесплатно"`, если цена равна нулю;
- иначе цену и валюту через пробел, например `"499 руб."` или `"10 $"`.

Функция ничего не печатает - только возвращает.

```python
# ТВОЙ КОД
def format_price(price, currency="руб."):
    if price == 0:
        return "бесплатно"
    else:
        #result = str(price) + " " + currency
        result = f"{price} {currency}" # "499 руб."
        return result


print(format_price(499))
print(format_price(0))
print(format_price(10, currency="$"))
```

```python
# Проверка
assert format_price(499) is not None, "функция вернула None - проверь, что там return, а не print"
assert isinstance(format_price(499), str), "функция должна возвращать строку"
assert format_price(499) == "499 руб.", f"ожидалось '499 руб.', а получилось {format_price(499)!r}"
assert format_price(0) == "бесплатно", f"ожидалось 'бесплатно', а получилось {format_price(0)!r}"
assert format_price(10, currency="$") == "10 $", "валюта должна подставляться из параметра currency"
assert format_price(10, "€") == "10 €", "currency можно передать и без имени, вторым аргументом"
assert format_price(0, "$") == "бесплатно", "ноль - всегда 'бесплатно', в любой валюте"
print("OK - задача 5 решена")
```
